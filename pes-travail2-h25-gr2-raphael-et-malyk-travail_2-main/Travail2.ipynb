```python
import pandas as pd
import matplotlib.pyplot as plt 


def analyse_co2_selon_le_jour(jour_en_francais):
    traduction = {
        "lundi": "Monday",
        "mardi": "Tuesday",
        "mercredi": "Wednesday",
        "jeudi": "Thursday",
        "vendredi": "Friday",
        "samedi": "Saturday",
        "dimanche": "Sunday"
    }

    jour_en_francais = jour_en_francais.lower()
    if jour_en_francais not in traduction:
        raise ValueError("Le jour que vous avez rentrer ne correspond pas à l'un des suivants : lundi, mardi, mercredi, jeudi, vendredi, samedi, dimanche.")
    
    jour_en_anglais = traduction[jour_en_francais]
    df = pd.read_csv("co2.csv", sep=";")
    df["date"] = pd.to_datetime(df["date"])
    df["dayofweek"] = df["date"].dt.strftime("%A")
    df["hour"]= df["date"].dt.hour

    donnees_filtrees = df[df["dayofweek"]== jour_en_anglais]

    regroupement_des_donnees = (
        donnees_filtrees.groupby("hour")["co2"].mean().reset_index()
    )

#le graphique a faire 

plt.figure(figsize=(10, 6))
    plt.plot(regroupement_des_donnees["hour"], regroupement_des_donnees["co2"], marker="o")

plt.title(f"quantité moyenne de CO2 le {jour_en_francais}")
plt.xlabel("Heure", fontsize=12)
plt.ylabel("CO2 (ppm)", fontsize=12)
plt.legend([jour_en_francais])
plt.grid(True)
   



    



                                
```

### Mise en contexte
Voici un exemple de code fonctionnel qui permet l'entraînement d'un réseau de neurones pour prédire si une tumeur est bénigne ou maligne(cancéreuse). Les données proviennent du site web : https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic. Initialement, ces données avaient été compilées par l'université du Wisconsin en 1995. Ce sont donc des données authentiques.

Dans ce numéro, le but est de vous initier sommairement à l'apprentissage machine d'une manière concrète et de voir les possibilités que cette approche offre.

Dans les fichiers de ce travail, vous retrouverez 2 fichiers :
* wdbc.names : ce fichier contient les informations sur les données. Les créateurs, le nom des colonnes pour les données ainsi que d'autres informations importantes.
* wdbc.data : ce fichier contient les données sous la forme de csv.

Les questions que nous vous invitons à poser doivent être : "Pourrais-tu m'expliquer ce que fait cette ligne de code : ...".
```python
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

#Lecture des données
donnees = pd.read_csv('wdbc.data')

# Prétraitement des données
# On retire l'id et la prédiction des données d'entrainement
caracteristique = donnees.drop(columns=[donnees.columns[0], donnees.columns[1]]) 
# On définit la colonne qui donne les prédictions
classification = donnees[donnees.columns[1]]

# Transforme nos classifications en nombre (B -> 0, M -> 1)
encoder = LabelEncoder()
classification = encoder.fit_transform(classification)

# Ces étapes vont transformer les données pour les mettre à l'échelle avec une moyenne de 0 et un écart-type de 1
# En ramenant les caractéristiques sur la même échelle, on évite qu'une caractéristique avec de plus grandes valeurs soit considérée plus importante par le modèle
scaler = StandardScaler()
caracteristique = scaler.fit_transform(caracteristique)

# On convertit en tensor. Un tensor est une matrice.
caracteristique = torch.tensor(caracteristique, dtype=torch.float32)
classification = torch.tensor(classification, dtype=torch.float32).view(-1, 1)

# On divise nos données avec 80 en entrainement et 20 en test.
donnees_caracteristique_entrainement, donnees_caracteristique_test, donnees_classification_entrainement, donnees_classification_test = train_test_split(caracteristique, classification, test_size=0.20, random_state=15)

# On définit notre réseau de neurones.
class Model(nn.Module):
    def __init__(self, input_size):
        super(Model, self).__init__()
        self.fc1 = nn.Linear(input_size, 12)
        self.fc2 = nn.Linear(12, 32)
        self.fc3 = nn.Linear(32, 32)
        self.fc4 = nn.Linear(32, 32)
        self.fc5 = nn.Linear(32, 32)
        self.fc6 = nn.Linear(32, 1)  # On a 1 sortie donc c'est une classification binaire. Le chiffre ici est habituellement le nombre de classes.

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = torch.relu(self.fc3(x))
        x = torch.relu(self.fc4(x))
        x = torch.relu(self.fc5(x))
        x = torch.sigmoid(self.fc6(x))  
        return x

# Initialise le modèle, la fonction de coût et l'optimiseur.
model = Model(donnees_caracteristique_entrainement.shape[1])
fonction_de_cout = nn.BCELoss()  # Binary Cross Entropy Loss (fonction de coût) pour classification binaire.
fonction_optimisation = optim.Adam(model.parameters(), lr=0.001) # Fonction qui met à jour le poids des neurones. Adam est basée sur la descente de radiants.

def tester_model():
    model.eval()
    with torch.no_grad():
        predictions = model(donnees_caracteristique_test)
        classes_predites = (predictions >= 0.5).float()  # 0 à 0.49 => 0 et 0.5 et plus => 1
        precision = (classes_predites == donnees_classification_test).float().mean()
        print(f"Test précision: {precision.item() * 100:.2f}%")

def entrainer_model():     
    # Entrainement
    num_epochs = 1000 
    for epoch in range(num_epochs):
        model.train()

        # Forward
        sorties = model(donnees_caracteristique_entrainement)
        cout = fonction_de_cout(sorties, donnees_classification_entrainement)

        # Backward 
        fonction_optimisation.zero_grad()
        cout.backward()
        fonction_optimisation.step()

        if (epoch+1) % 10 == 0:
            print(f"Epoch [{epoch+1}/{num_epochs}], Fonction de coût: {cout.item():.4f}")
            tester_model()
            
entrainer_model()
```
